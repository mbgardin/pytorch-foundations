# 01 - Tensors: PyTorch's basic data container

**Goals:** create tensors; track shapes, dtypes, and devices; use indexing, broadcasting, and matrix products.

A tensor is a multidimensional array, much like a NumPy array. Run cells in order in a fresh kernel; each notebook supplies its own setup.

```python
import numpy as np
import torch

torch.manual_seed(486)
```

## Create and inspect a tensor

Read `(2, 3)` as two rows and three columns. In later notebooks, rows will represent observations and columns will represent features.

```python
x = torch.tensor([[1., 2., 3.], [4., 5., 6.]])
print(x)
print("shape:", x.shape, "dimensions:", x.ndim)
print("dtype:", x.dtype, "device:", x.device)
```

Indexing with a single integer removes that axis. Slicing the rows with `:` keeps all rows while selecting a column.

```python
print("first row:", x[0], "shape:", x[0].shape)
print("last column:", x[:, -1], "shape:", x[:, -1].shape)
```

## Elementwise operations and broadcasting

`*` multiplies corresponding entries; it preserves the shape here.

```python
print("elementwise square:\n", x * x)
```

A length-three vector broadcasts across the rows of a `(2, 3)` tensor. Predict which value is added to each column.

```python
offset = torch.tensor([10., 20., 30.])
print("broadcast addition:\n", x + offset)
```

## Matrix multiplication and reductions

For `@`, the inner dimensions must match: `(2, 3) @ (3, 2)` gives `(2, 2)`. Each output entry is a row-column dot product.

```python
weights = torch.tensor([[1., 0.], [0., 1.], [1., 1.]])
print("matrix product:\n", x @ weights)
```

`mean(dim=0)` averages over rows, leaving one mean per column. `mean()` with no dimension averages all entries to a scalar.

```python
print("column means:", x.mean(dim=0))
print("overall mean:", x.mean())
```

## Reshape and transpose

Both results below have shape `(3, 2)`, but contain different arrangements. `reshape` preserves the sequence of elements in this example; `.T` swaps the two axes.

```python
print("reshape:\n", x.reshape(3, 2))
print("transpose:\n", x.T)
```

Selecting one row gives shape `(3,)`. `unsqueeze(0)` inserts an axis so a single observation still has a batch dimension, `(1, 3)`.

```python
print("one row:", x[0].shape)
print("one-row batch:", x[0].unsqueeze(0).shape)
```

## Dtypes, NumPy, and devices

Use floating-point inputs for these models and integer class indices for classification labels.

```python
labels = torch.tensor([0, 1], dtype=torch.long)
print("feature dtype:", x.dtype, "label dtype:", labels.dtype)
```

`torch.from_numpy` shares memory with this NumPy array. Change the array and observe the tensor; use `.clone()` for an independent tensor copy.

```python
arr = np.array([1, 2, 3], dtype=np.float32)
t = torch.from_numpy(arr)
arr[0] = 10
print("shared values:", arr, t)
print("back to NumPy:", t.numpy())
```

These examples run on the CPU. In a model computation, inputs and parameters need compatible dtypes and the same device; see the [setup guide](../guides/01-setup.md) for accelerators.

```python
device = torch.device("cpu")
print("on device:", x.to(device).device)
```
